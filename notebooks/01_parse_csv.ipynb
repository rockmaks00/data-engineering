1. Считать данные из файла `test_dataset.csv`.

```python
import pandas as pd
from pathlib import Path

path = Path("/home/jovyan/data/test_dataset.csv")

df = pd.read_csv(path, sep="|", decimal=".", parse_dates=["load_date"])

# приводим к числовым типам то, что не преобразовалось автоматически
for col in ["lat", "lon"]:
    df[col] = (
        df[col]
        .str.replace(",", ".", regex=False)
        .astype(float)
    )

df.head(10)
```

2. Вывести основные статистические метрики по каждому числовому столбцу:

   * количество записей;
   * минимальное значение (`min`);
   * максимальное значение (`max`);
   * среднее значение (`mean`);
   * стандартное отклонение (`std`).

```python
numeric_df = df.select_dtypes(include="number")

stats = numeric_df.describe().T[["count", "min", "max", "mean", "std"]]

# для читаемости округляем до 2 знаков и приводим к классическому виду
stats.style.format(
    lambda x: f"{x:,.2f}".replace(",", " ").replace(".", ",")
)
```
